```python
import sqlalchemy as db

engine = db.create_engine(
    'mysql+mysqlconnector://root:root@127.0.0.1:3310/retail_db?charset=utf8mb4',
    pool_pre_ping=True,
    future=True
)
with engine.connect() as conn:
    df_departments = pd.read_sql_table('departments', con=conn, schema='retail_db')
df_departments
```

```python
import pandas as pd
from sqlalchemy import text

with engine.connect() as conn:
    print("DB actual:", conn.exec_driver_sql("SELECT DATABASE();").scalar())
    tablas = pd.read_sql(
        text("SELECT table_name FROM information_schema.tables WHERE table_schema = DATABASE() ORDER BY 1"),
        conn
    )
tablas
```

```python
df_departments = pd.read_sql_query(text("SELECT * FROM departments"), con=engine)
df_departments
```
